```python
import pandas as pd
from pathlib import Path

HOME_PATH = Path("..")
```

```python
df = pd.read_csv(HOME_PATH / "data/dataset/AzharAli05_Resume_Screening_Dataset.csv")
df.head()
```

```python
df['description'].sample(150, random_state=0)
```

```python
# grab ten random jobs
jobs = df['description'].drop_duplicates().sample(n=150, random_state=0)
jobs
```

```python
df = df[df['description'].isin(jobs)]
df = df.groupby(['description', 'decision'], group_keys=False).apply(lambda df: df.sample(n=1, random_state=0)).reset_index(drop=True)
df
```

```python
# let's convert this to a "standard form"
new_dict = [
    {
        'description': d,
        'resume_accept': df[
            (df['description']==d) & (df['decision']==True)
        ]['resume'].item(),
        'resume_reject': df[
            (df['description']==d) & (df['decision']==False)
        ]['resume'].item(),
    } for d in df['description'].unique()
]
df_2 = pd.DataFrame(new_dict)
df_2
```

```python
df_2.to_csv(HOME_PATH / "data/dataset/AzharAli05_Resume_subset150.csv", index=False)
```
